```python
import napari
import numpy as np
from tifffile import TiffFile
```

```python
viewer = napari.Viewer()
napari.run()
```

```python
# scale all layers in z by 5

z_scale = 5
for layer in viewer.layers:
# for layer in list(viewer.layers)[-6:]: 
    layer.scale = [z_scale, 1, 1]
```

```python
# scale a specific layer

# layer = viewer.layers['Stack_Run04_125130_MMStack_Pos0.ome']

layer = viewer.layers['Stack_Run04_T003_deconvolved_DL2_RL80.ome']
layer.scale = [z_scale, 1, 1]
```

```python
# layer = viewer.layers['Stack_Run26_135705_MMStack_Pos0.ome']
layer = viewer.layers['Stack_Run04_125130_MMStack_Pos0.ome']
layer.data = layer.data[::-1]
data = layer.data
print(data.shape)
```

```python
# cut fast piezo data to only include 1 timepoint

n_timepoints = 6
t_index = 3  # the timepoint you want (1-indexed)

chunk_size = data.shape[0] // n_timepoints

start = (t_index - 1) * chunk_size
end = t_index * chunk_size

cropped_data = data[start:end]

viewer.add_image(
    cropped_data,
    name=layer.name + f"_t{t_index}",
    scale=layer.scale,
)
```

```python
# flip z data

toflip = viewer.layers[f'Stack_Run04_125130_MMStack_Pos0.ome_t{t_index}']
# toflip = viewer.layers['Stack_Run26_135705_MMStack_Pos0.ome_t4']
flipped_data = toflip.data[::-1]

viewer.add_image(
    flipped_data,
    name=layer.name + "_zflipped",
    scale=layer.scale,
)
```
